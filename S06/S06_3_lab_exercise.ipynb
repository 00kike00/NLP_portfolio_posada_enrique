````python
# Q1 (1 pt): Sentence splitting (regex + NLTK)
# - Use regex to protect acronyms like U.P.C. so they don't break sentence boundaries.
# - Then use nltk.sent_tokenize.
#
# Return: sentences (list of strings)

# TODO: implement protect_acronym_dots and restore_acronym_dots (or equivalent)
# TODO: apply sent_tokenize
def protect_acronym_dots(text):
    # We replace the dots of acronyms with ohter character to protect them
    return re.sub(r'\b(?:[A-Z]\.){2,}', lambda m: m.group().replace('.', '#'), text)


def restore_acronym_dots(text):
    return  text.replace("#", ".")

dot_protected = protect_acronym_dots(text)
sentences = sent_tokenize(dot_protected)
for i in range(len(sentences)):
    sentences[i] = restore_acronym_dots(sentences[i])


print(sentences)
````

````python
# Q8 (2 pt): Simple BPE (Byte Pair Encoding) on a tiny corpus
corpus = "low lower newest widest"

# Requirements:
# - Represent each word as characters + </w>
# - Compute pair frequencies (weighted by word frequency)
# - Merge most frequent pair
# - Do at least 5 merges (or stop if no pairs)
#
# Deliver:
# - merges: list of merges in order
# - final segmented version of each word

merges = None

def get_vocab_from_corpus(corpus):
    vocab = defaultdict(int)
    for word in corpus.split():
        token = tuple(list(word) + ['</w>'])
        vocab[token] += 1
    return vocab
v = get_vocab_from_corpus(corpus)
print(v)

def get_pair_frequencies(vocab):
    # Count how often each adjacent pair appears
    pairs = defaultdict(int)
    for word_tokens, freq in vocab.items():
        for a, b in zip(word_tokens, word_tokens[1:]):
            pairs[(a, b)] += freq
    return pairs

pairs = get_pair_frequencies(v)
print(pairs)

def merge_pair_in_vocab(pair, vocab):
    # Replace every pair that matches the given with the merged token in all words
    new_vocab = {}
    a, b = pair
    merged = a + b
    for word_tokens, freq in vocab.items():
        new_tokens = []
        i = 0
        while i < len(word_tokens):
            if i < len(word_tokens) - 1 and word_tokens[i] == a and word_tokens[i+1] == b:
                new_tokens.append(merged)
                i += 2
            else:
                new_tokens.append(word_tokens[i])
                i += 1
        new_vocab[tuple(new_tokens)] = freq
    return new_vocab

merged = merge_pair_in_vocab(('l', 'o'), v)
print(merged)

# TODO: implement BPE helper functions:
# - get_vocab_from_corpus
# - get_pair_frequencies
# - merge_pair_in_vocab

# print(merges)
````

````python
# Q9 (1 pt): Metrics — Accuracy, Precision, Recall, F1
# Invent a confusion matrix (TP, FP, FN, TN) and compute metrics.
# Explain each formula briefly in comments.

TP = 50
FP = 10
FN = 20
TN = 120

# Calculate metrics
accuracy = (TP + TN) / (TP + FP + FN + TN) # We divide the correct classified counts btween the sum of everything
precision = TP / (TP + FP) # We divide the correctly classified True between the sum of every classified True
recall = TP / (TP + FN) # # We divide the correctly classified True between the sum of every True
f1 = 2 * precision * recall / (precision + recall) # We divide the product of 2 mulitplied by precicison multiplied by recall and divide it between the sum of precision + recall

print(accuracy, precision, recall, f1)
````
